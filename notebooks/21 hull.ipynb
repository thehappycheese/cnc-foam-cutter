```python
import numpy as np
from airfoil.wing import auto_piecewise, auto_interpolate
import matplotlib.pyplot as plt
```

```python
upper = auto_piecewise([
    (200, auto_interpolate([
        (0,0),
        (200,50)
    ])),
    (400, lambda x: 50)
])

x = np.linspace(0,400)
y = upper(x)

plt.plot(x,y)

```
